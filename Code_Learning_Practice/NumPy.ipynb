# Initializing Different Types of Arrays

```python
# All 0s matrix
```

```python
np.zeros((2,3,3))
```

```python
# All 1s matris
np.ones((2,3,3), dtype= 'int32')
```

```python
# Any other number
np.full((2,2), 99)
```

```python
np.full((2,2), 99, dtype='float32')
```

```python
# Any other number (full_like)
np.full_like(a,4)
```

```python
np.full(a.shape,4)
```

```python
# Random decimal numbers
np.random.rand(4,2)
```

```python
# Random Integer Value
np.random.randint(1,5, size=(3,3))
```

```python
# The Identity Matrix
np.identity(5)
```

```python
#r= Repeat array multiple times
arr = np.array([[1,2,3]])
r1= np.repeat(arr,3, axis = 0)
print (r1)
```

```python
arr = np.array([[1,2,3]])
r1= np.repeat(arr,3, axis = 1)
print (r1)
```
